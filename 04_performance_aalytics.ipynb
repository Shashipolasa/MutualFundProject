```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

performance_metrics = pd.read_csv("performance_metrics.csv")

performance_metrics.head()
```

```python
top_cagr = performance_metrics.sort_values(
    "cagr",
    ascending=False
).head(10)

plt.figure(figsize=(10,5))
plt.bar(top_cagr["amfi_code"].astype(str), top_cagr["cagr"])

plt.title("Top 10 Funds by CAGR")
plt.xlabel("AMFI Code")
plt.ylabel("CAGR")

plt.xticks(rotation=45)
plt.show()
```

```python
top_sharpe = performance_metrics.sort_values(
    "sharpe_ratio",
    ascending=False
).head(10)

plt.figure(figsize=(10,5))
plt.bar(top_sharpe["amfi_code"].astype(str),
        top_sharpe["sharpe_ratio"])

plt.title("Top 10 Funds by Sharpe Ratio")
plt.xlabel("AMFI Code")
plt.ylabel("Sharpe Ratio")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    performance_metrics["annual_volatility"],
    performance_metrics["annual_return"]
)

plt.title("Risk vs Return")
plt.xlabel("Annual Volatility")
plt.ylabel("Annual Return")

plt.show()
```

```python
plt.savefig("top10_cagr.png", bbox_inches="tight")
```
