```python
import pandas as pd
import numpy as np

performance_metrics = pd.read_csv("performance_metrics.csv")
var_df = pd.read_csv("var_metrics.csv")

ranking_df = performance_metrics.merge(
    var_df,
    on="amfi_code"
)

ranking_df.head()
```

```python
def risk_category(volatility):
    if volatility < 0.10:
        return "Low Risk"
    elif volatility < 0.20:
        return "Medium Risk"
    else:
        return "High Risk"

ranking_df["risk_category"] = ranking_df["annual_volatility"].apply(
    risk_category
)

ranking_df[
    ["amfi_code","annual_volatility","risk_category"]
].head()
```

```python
ranking_df["recommendation"] = np.where(
    (ranking_df["sharpe_ratio"] > 1) &
    (ranking_df["cagr"] > 0.15),
    "Recommended",
    "Not Recommended"
)

ranking_df[
    ["amfi_code","sharpe_ratio","cagr","recommendation"]
].head()
```

```python
recommended_funds = ranking_df[
    ranking_df["recommendation"] == "Recommended"
]

recommended_funds = recommended_funds.sort_values(
    by="sharpe_ratio",
    ascending=False
)

recommended_funds[
    ["amfi_code",
     "sharpe_ratio",
     "cagr",
     "risk_category"]
].head(10)
```

```python
recommended_funds.to_csv(
    "recommended_funds.csv",
    index=False
)

print("Recommended Funds Saved Successfully")
```
